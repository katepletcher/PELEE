```python
# split_type = 'simple'
split_type = 'mcquery'
```

```python
vars = ["pot", "run", "subRun"]
vars_query = ["run", "sub", "nu_pdg", "ccnc", "mcf_pass_ncpi0", "category"]

NUEQUERY = '(abs(nu_pdg)==12 & ccnc==0)'
NCPI0QUERY = '(mcf_pass_ncpi0==1)'
MCQUERY = '((abs(nu_pdg)==12 & ccnc==0) | mcf_pass_ncpi0==1)'

# Run 4c
nu4c_subRun = u_nu4c.arrays(vars, library='pd')
nu4c_nueSelFilter = u_nu4c_nufilter.arrays(vars_query, library='pd')
nu4c_duplicates = pd.merge(nu4c_subRun, nu4c_nueSelFilter, left_on=["run","subRun"], right_on=["run","sub"], how="left")
nu4c = nu4c_duplicates.drop_duplicates(subset=['subRun','pot'])
nue4c_subRun = u_nue4c.arrays(vars, library='pd')
nue4c_nueSelFilter = u_nue4c_nufilter.arrays(vars_query, library='pd')
nue4c_duplicates = pd.merge(nue4c_subRun, nue4c_nueSelFilter, left_on=["run","subRun"], right_on=["run","sub"], how="left")
nue4c = nue4c_duplicates.drop_duplicates(subset=['subRun','pot'])
dirt4c = u_dirt4c.arrays(vars, library='pd')
ncpi04c = u_ncpi04c.arrays(vars, library='pd')

# Run 4d
nu4d_subRun = u_nu4d.arrays(vars, library='pd')
nu4d_nueSelFilter = u_nu4d_nufilter.arrays(vars_query, library='pd')
nu4d_duplicates = pd.merge(nu4d_subRun, nu4d_nueSelFilter, left_on=["run","subRun"], right_on=["run","sub"], how="left")
nu4d = nu4d_duplicates.drop_duplicates(subset=['subRun','pot'])
nue4d_subRun = u_nue4d.arrays(vars, library='pd')
nue4d_nueSelFilter = u_nue4d_nufilter.arrays(vars_query, library='pd')
nue4d_duplicates = pd.merge(nue4d_subRun, nue4d_nueSelFilter, left_on=["run","subRun"], right_on=["run","sub"], how="left")
nue4d = nue4d_duplicates.drop_duplicates(subset=['subRun','pot'])
dirt4d = u_dirt4d.arrays(vars, library='pd')
ncpi04d = u_ncpi04d.arrays(vars, library='pd')

# Run 5
nu5_subRun = u_nu5.arrays(vars, library='pd')
nu5_nueSelFilter = u_nu5_nufilter.arrays(vars_query, library='pd')
nu5_duplicates = pd.merge(nu5_subRun, nu5_nueSelFilter, left_on=["run","subRun"], right_on=["run","sub"], how="left")
nu5 = nu5_duplicates.drop_duplicates(subset=['subRun','pot'])
nue5_subRun = u_nue5.arrays(vars, library='pd')
nue5_nueSelFilter = u_nue5_nufilter.arrays(vars_query, library='pd')
nue5_duplicates = pd.merge(nue5_subRun, nue5_nueSelFilter, left_on=["run","subRun"], right_on=["run","sub"], how="left")
nue5 = nue5_duplicates.drop_duplicates(subset=['subRun','pot'])
dirt5 = u_dirt5.arrays(vars, library='pd')
ncpi05 = u_ncpi05.arrays(vars, library='pd')

if split_type == 'simple':
    train_nu4c, test_nu4c = train_test_split(nu4c, test_size=0.5, random_state=1990)
    train_nue4c, test_nue4c = train_test_split(nue4c, test_size=0.5, random_state=1990)
    train_dirt4c, test_dirt4c = train_test_split(dirt4c, test_size=0.5, random_state=1990)
    train_ncpi04c, test_ncpi04c = train_test_split(ncpi04c, test_size=0.5, random_state=1990)

    train_nu4d, test_nu4d = train_test_split(nu4d, test_size=0.5, random_state=1990)
    train_nue4d, test_nue4d = train_test_split(nue4d, test_size=0.5, random_state=1990)
    train_dirt4d, test_dirt4d = train_test_split(dirt4d, test_size=0.5, random_state=1990)
    train_ncpi04d, test_ncpi04d = train_test_split(ncpi04d, test_size=0.5, random_state=1990)

    train_nu5, test_nu5 = train_test_split(nu5, test_size=0.5, random_state=1990)
    train_nue5, test_nue5 = train_test_split(nue5, test_size=0.5, random_state=1990)
    train_dirt5, test_dirt5 = train_test_split(dirt5, test_size=0.5, random_state=1990)
    train_ncpi05, test_ncpi05 = train_test_split(ncpi05, test_size=0.5, random_state=1990)

if split_type == 'mcquery':
    nu4c_ccnue = nu4c.query(NUEQUERY)
    nu4c_ncpi0 = nu4c.query(NCPI0QUERY)
    nu4c_no_ccnue_ncpi0 = nu4c.query('~'+MCQUERY)

    train_nu4c, test_nu4c = train_test_split(nu4c_no_ccnue_ncpi0, test_size=0.75, random_state=1990)
    train_nue4c_truth, test_nue4c = train_test_split(nue4c, test_size=0.75, random_state=1990)
    train_nue4c = pd.concat([train_nue4c_truth, nu4c_ccnue], ignore_index=True)
    train_dirt4c, test_dirt4c = train_test_split(dirt4c, test_size=0.75, random_state=1990)
    train_ncpi04c_truth, test_ncpi04c = train_test_split(ncpi04c, test_size=0.75, random_state=1990)
    train_ncpi04c = pd.concat([train_ncpi04c_truth, nu4c_ncpi0], ignore_index=True)

    nu4d_ccnue = nu4d.query(NUEQUERY)
    nu4d_ncpi0 = nu4d.query(NCPI0QUERY)
    nu4d_no_ccnue_ncpi0 = nu4d.query('~'+MCQUERY)

    train_nu4d, test_nu4d = train_test_split(nu4d_no_ccnue_ncpi0, test_size=0.75, random_state=1990)
    train_nue4d_truth, test_nue4d = train_test_split(nue4d, test_size=0.75, random_state=1990)
    train_nue4d = pd.concat([train_nue4d_truth, nu4d_ccnue], ignore_index=True)
    train_dirt4d, test_dirt4d = train_test_split(dirt4d, test_size=0.75, random_state=1990)
    train_ncpi04d_truth, test_ncpi04d = train_test_split(ncpi04d, test_size=0.75, random_state=1990)
    train_ncpi04d = pd.concat([train_ncpi04d_truth, nu4d_ncpi0], ignore_index=True)

    nu5_ccnue = nu5.query(NUEQUERY)
    nu5_ncpi0 = nu5.query(NCPI0QUERY)
    nu5_no_ccnue_ncpi0 = nu5.query('~'+MCQUERY)

    train_nu5, test_nu5 = train_test_split(nu5_no_ccnue_ncpi0, test_size=0.75, random_state=1990)
    train_nue5_truth, test_nue5 = train_test_split(nue5, test_size=0.75, random_state=1990)
    train_nue5 = pd.concat([train_nue5_truth, nu5_ccnue], ignore_index=True)
    train_dirt5, test_dirt5 = train_test_split(dirt5, test_size=0.75, random_state=1990)
    train_ncpi05_truth, test_ncpi05 = train_test_split(ncpi05, test_size=0.75, random_state=1990)
    train_ncpi05 = pd.concat([train_ncpi05_truth, nu5_ncpi0], ignore_index=True)
```
