```python
# split_type = 'simple'
split_type = 'mcquery'
```

```python
vars = ["pot", "run", "subRun"]
vars_query = ["run", "sub", "nu_pdg", "ccnc", "mcf_pass_ncpi0", "category"]

NUEQUERY = '(abs(nu_pdg)==12 & ccnc==0)'
NCPI0QUERY = '(mcf_pass_ncpi0==1)'
MCQUERY = '((abs(nu_pdg)==12 & ccnc==0) | mcf_pass_ncpi0==1)'

# Run 4a
nu4a_subRun = u_nu4a.arrays(vars, library='pd')
nu4a_nueSelFilter = u_nu4a_nufilter.arrays(vars_query, library='pd')
nu4a_duplicates = pd.merge(nu4a_subRun, nu4a_nueSelFilter, left_on=["run","subRun"], right_on=["run","sub"], how="left")
nu4a = nu4a_duplicates.drop_duplicates(subset=['subRun','pot'])
nue4a_subRun = u_nue4a.arrays(vars, library='pd')
nue4a_nueSelFilter = u_nue4a_nufilter.arrays(vars_query, library='pd')
nue4a_duplicates = pd.merge(nue4a_subRun, nue4a_nueSelFilter, left_on=["run","subRun"], right_on=["run","sub"], how="left")
nue4a = nue4a_duplicates.drop_duplicates(subset=['subRun','pot'])
dirt4a = u_dirt4a.arrays(vars, library='pd')
ncpi04a = u_ncpi04a.arrays(vars, library='pd')

# Run 4b
nu4b_subRun = u_nu4b.arrays(vars, library='pd')
nu4b_nueSelFilter = u_nu4b_nufilter.arrays(vars_query, library='pd')
nu4b_duplicates = pd.merge(nu4b_subRun, nu4b_nueSelFilter, left_on=["run","subRun"], right_on=["run","sub"], how="left")
nu4b = nu4b_duplicates.drop_duplicates(subset=['subRun','pot'])
nue4b_subRun = u_nue4b.arrays(vars, library='pd')
nue4b_nueSelFilter = u_nue4b_nufilter.arrays(vars_query, library='pd')
nue4b_duplicates = pd.merge(nue4b_subRun, nue4b_nueSelFilter, left_on=["run","subRun"], right_on=["run","sub"], how="left")
nue4b = nue4b_duplicates.drop_duplicates(subset=['subRun','pot'])
dirt4b = u_dirt4b.arrays(vars, library='pd')
ncpi04b = u_ncpi04b.arrays(vars, library='pd')

# Run 4c
nu4c_subRun = u_nu4c.arrays(vars, library='pd')
nu4c_nueSelFilter = u_nu4c_nufilter.arrays(vars_query, library='pd')
nu4c_duplicates = pd.merge(nu4c_subRun, nu4c_nueSelFilter, left_on=["run","subRun"], right_on=["run","sub"], how="left")
nu4c = nu4c_duplicates.drop_duplicates(subset=['subRun','pot'])
nue4c_subRun = u_nue4c.arrays(vars, library='pd')
nue4c_nueSelFilter = u_nue4c_nufilter.arrays(vars_query, library='pd')
nue4c_duplicates = pd.merge(nue4c_subRun, nue4c_nueSelFilter, left_on=["run","subRun"], right_on=["run","sub"], how="left")
nue4c = nue4c_duplicates.drop_duplicates(subset=['subRun','pot'])
dirt4c = u_dirt4c.arrays(vars, library='pd')
ncpi04c = u_ncpi04c.arrays(vars, library='pd')

if split_type == 'simple':
    train_nu4a, test_nu4a = train_test_split(nu4a, test_size=0.75, random_state=1990)
    train_nue4a, test_nue4a = train_test_split(nue4a, test_size=0.75, random_state=1990)
    train_dirt4a, test_dirt4a = train_test_split(dirt4a, test_size=0.75, random_state=1990)
    train_ncpi04a, test_ncpi04a = train_test_split(ncpi04a, test_size=0.75, random_state=1990)

    train_nu4b, test_nu4b = train_test_split(nu4b, test_size=0.75, random_state=1990)
    train_nue4b, test_nue4b = train_test_split(nue4b, test_size=0.75, random_state=1990)
    train_dirt4b, test_dirt4b = train_test_split(dirt4b, test_size=0.75, random_state=1990)
    train_ncpi04b, test_ncpi04b = train_test_split(ncpi04b, test_size=0.75, random_state=1990)

    train_nu4c, test_nu4c = train_test_split(nu4c, test_size=0.75, random_state=1990)
    train_nue4c, test_nue4c = train_test_split(nue4c, test_size=0.75, random_state=1990)
    train_dirt4c, test_dirt4c = train_test_split(dirt4c, test_size=0.75, random_state=1990)
    train_ncpi04c, test_ncpi04c = train_test_split(ncpi04c, test_size=0.75, random_state=1990)

if split_type == 'mcquery':
    nu4a_ccnue = nu4a.query(NUEQUERY)
    nu4a_ncpi0 = nu4a.query(NCPI0QUERY)
    nu4a_no_ccnue_ncpi0 = nu4a.query('~'+MCQUERY)

    train_nu4a, test_nu4a = train_test_split(nu4a_no_ccnue_ncpi0, test_size=0.75, random_state=1990)
    train_nue4a_truth, test_nue4a = train_test_split(nue4a, test_size=0.75, random_state=1990)
    train_nue4a = pd.concat([train_nue4a_truth, nu4a_ccnue], ignore_index=True)
    train_dirt4a, test_dirt4a = train_test_split(dirt4a, test_size=0.75, random_state=1990)
    train_ncpi04a_truth, test_ncpi04a = train_test_split(ncpi04a, test_size=0.75, random_state=1990)
    train_ncpi04a = pd.concat([train_ncpi04a_truth, nu4a_ncpi0], ignore_index=True)

    nu4b_ccnue = nu4b.query(NUEQUERY)
    nu4b_ncpi0 = nu4b.query(NCPI0QUERY)
    nu4b_no_ccnue_ncpi0 = nu4b.query('~'+MCQUERY)

    train_nu4b, test_nu4b = train_test_split(nu4b_no_ccnue_ncpi0, test_size=0.75, random_state=1990)
    train_nue4b_truth, test_nue4b = train_test_split(nue4b, test_size=0.75, random_state=1990)
    train_nue4b = pd.concat([train_nue4b_truth, nu4b_ccnue], ignore_index=True)
    train_dirt4b, test_dirt4b = train_test_split(dirt4b, test_size=0.75, random_state=1990)
    train_ncpi04b_truth, test_ncpi04b = train_test_split(ncpi04b, test_size=0.75, random_state=1990)
    train_ncpi04b = pd.concat([train_ncpi04b_truth, nu4b_ncpi0], ignore_index=True)

    nu4c_ccnue = nu4c.query(NUEQUERY)
    nu4c_ncpi0 = nu4c.query(NCPI0QUERY)
    nu4c_no_ccnue_ncpi0 = nu4c.query('~'+MCQUERY)

    train_nu4c, test_nu4c = train_test_split(nu4c_no_ccnue_ncpi0, test_size=0.75, random_state=1990)
    train_nue4c_truth, test_nue4c = train_test_split(nue4c, test_size=0.75, random_state=1990)
    train_nue4c = pd.concat([train_nue4c_truth, nu4c_ccnue], ignore_index=True)
    train_dirt4c, test_dirt4c = train_test_split(dirt4c, test_size=0.75, random_state=1990)
    train_ncpi04c_truth, test_ncpi04c = train_test_split(ncpi04c, test_size=0.75, random_state=1990)
    train_ncpi04c = pd.concat([train_ncpi04c_truth, nu4c_ncpi0], ignore_index=True)
```
